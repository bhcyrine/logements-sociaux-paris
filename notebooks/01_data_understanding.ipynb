```python
dupe_ids = df[df.duplicated('id_livraison', keep=False)].sort_values('id_livraison')
print(f"{dupe_ids['id_livraison'].nunique()} identifiants distincts apparaissent plus d'une fois "
      f"({len(dupe_ids)} lignes au total)\n")

print(dupe_ids[['id_livraison', 'adresse_programme', 'annee', 'mode_real',
                 'nb_logmt_total']].to_string())

```

```python
# Indiquer si les lignes dupliquées partagent la même adresse et la même année (probablement un vrai doublon)
# ou si elles diffèrent (probablement une livraison distincte ou réalisée en plusieurs phases)
dupe_ids['same_addr_year'] = dupe_ids.duplicated(
    subset=['id_livraison', 'adresse_programme', 'annee'], keep=False
)
print("\nLignes où l'adresse et l'année sont identiques pour le même id_livraison "
      "(probablement de vrais doublons) :")
print(dupe_ids[dupe_ids['same_addr_year']])

print("\nLignes où l'adresse ou l'année diffère (probablement des livraisons distinctes légitimes) :")
print(dupe_ids[~dupe_ids['same_addr_year']])
```

```python
## Vérification de cohérence : un même programme (id_livraison) 
## doit avoir un seul mode_real, sinon l'agrégation par mode serait faussée

incoherent = df.groupby('id_livraison')['mode_real'].nunique()
n_incoherent = (incoherent > 1).sum()
print(f"Programmes avec mode_real incohérent entre leurs lignes : {n_incoherent}")

if n_incoherent > 0:
    ids_incoherents = incoherent[incoherent > 1].index
    print(df[df['id_livraison'].isin(ids_incoherents)]
          [['id_livraison', 'adresse_programme', 'annee', 'mode_real']])
```

```python
## 9ter. Test plus robuste : comparer le début de l'adresse (moins sensible 
## aux petites variations de texte) plutôt que l'égalité stricte

for id_ in ids_incoherents:
    sub = df[df['id_livraison'] == id_]
    adresses = sub['adresse_programme'].unique()
    # Compare seulement les 15 premiers caractères, normalisés
    prefixes = set(a.strip().upper()[:15] for a in adresses)
    same_prefix = len(prefixes) == 1
    flag = "OK (adresse similaire)" if same_prefix else "⚠️ SUSPECT (adresses vraiment différentes)"
    print(f"id_livraison {id_} — {flag}")
    if not same_prefix:
        for a in adresses:
            print(f"   -> {a}")
```

## Conclusions — Data Understanding

- **Négatifs** (`nb_plai`/`nb_plus`/`nb_logmt_total` < 0) : quelques lignes isolées, 
  probablement des corrections de déclarations antérieures. Impact négligeable 
  (2 lignes / 4174). Décision : exclues du dataset nettoyé.

- **id_livraison dupliqués** : NE SONT PAS des doublons. Ils représentent des tranches 
  de financement ou des phases de livraison distinctes d'un même programme. 
  Décision : conservées telles quelles, id_livraison n'est pas une clé unique par ligne.

- **id_livraison avec mode_real incohérent entre ses lignes** : 7 cas détectés sur le 
  dataset brut (10 lignes). Test de similarité d'adresse (préfixe normalisé) : 
  6 cas sont des tranches/phases légitimes du même programme (même adresse, à des 
  variations d'écriture près). 1 cas (id_livraison 2002196) présente deux adresses 
  réellement différentes (5-7 RUE DU SAHEL vs 1 RUE PAUL CRAMPEL, années 2012/2024), 
  suggérant une réutilisation de l'identifiant plutôt qu'un même programme. 
  Comme l'analyse se fait au niveau ligne (chaque ligne porte son propre mode_real 
  et ses comptes de financement), ceci n'affecte pas l'agrégation par mode_real. 
  Aucune action de nettoyage nécessaire ; documenté pour ne pas traiter id_livraison 
  comme identifiant de projet fiable.

- **bs = "(vide)"** : placeholder de valeur manquante (109 lignes), non utilisé dans 
  l'analyse principale. Pas d'action nécessaire.

- **Outlier nb_logmt_total = 1332** : légitime, cohérent avec les gros programmes 
  d'acquisition conventionnement (rachat de parc immobilier existant, ex. Porte de 
  Montreuil). Conservé.

- **nature_programme fortement corrélé à mode_real** (ex : résidences étudiantes 
  81% en acquisition réhabilitation, EHPAD 92% en acquisition conventionnement) : 
  facteur de confusion potentiel pour l'analyse principale. Décision : analyse 
  centrée sur `nature_programme == "logement familial"` (3551 programmes), avec 
  vérification de robustesse sur le dataset complet.
