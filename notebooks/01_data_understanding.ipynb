## 6. Investigation — valeurs négatives

Quelques lignes présentent des comptages négatifs, ce qui est impossible pour un nombre de logements.


```python
neg_cols = ['nb_logmt_total', 'nb_plai', 'nb_plus', 'nb_pluscd', 'nb_pls']
neg_mask = (df[neg_cols] < 0).any(axis=1)
print(f"{neg_mask.sum()} lignes avec au moins un nombre négatif\n")
neg_rows = df.loc[neg_mask, ['id_livraison', 'adresse_programme', 'annee', 'bs',
                              'mode_real', 'nb_logmt_total', 'nb_plai', 'nb_plus',
                              'nb_pluscd', 'nb_pls']]
print(neg_rows)
```

```python
print("Par mode_real:")
print(neg_rows["mode_real"].value_counts())
print("\nPar bs:")
print(neg_rows["bs"].value_counts())
print("\nPar annee:")
print(neg_rows["annee"].value_counts().sort_index())
```

```python
check = neg_rows.copy()
check['sum_categories'] = check[['nb_plai', 'nb_plus', 'nb_pluscd', 'nb_pls']].sum(axis=1)
check['matches_total'] = check['sum_categories'] == check['nb_logmt_total']
print("Vérification de cohérence (total == somme des sous-catégories) :")
print(check[['id_livraison', 'nb_logmt_total', 'sum_categories', 'matches_total']])
```

## 7. Investigation — `id_livraison` dupliqués

Vérifier si les identifiants qui apparaissent plusieurs fois sont de vrais doublons ou des tranches/phases légitimes du même programme.


```python
dupe_ids = df[df.duplicated('id_livraison', keep=False)].sort_values('id_livraison')
print(f"{dupe_ids['id_livraison'].nunique()} identifiants distincts apparaissent plus d'une fois "
      f"({len(dupe_ids)} lignes au total)\n")
print(dupe_ids[['id_livraison', 'adresse_programme', 'annee', 'mode_real',
                 'nb_logmt_total']].to_string())
```

```python
# Les lignes dupliquées partagent-elles la même adresse et la même année (vrai doublon)
# ou diffèrent-elles (livraison distincte / phases différentes) ?
dupe_ids['same_addr_year'] = dupe_ids.duplicated(
    subset=['id_livraison', 'adresse_programme', 'annee'], keep=False
)
print("Lignes où adresse + année sont identiques pour le même id_livraison "
      "(probablement de vrais doublons) :")
print(dupe_ids[dupe_ids['same_addr_year']])

print("\nLignes où l'adresse ou l'année diffère (probablement des livraisons distinctes légitimes) :")
print(dupe_ids[~dupe_ids['same_addr_year']])
```

## 8. Vérification — cohérence de `mode_real` au sein d'un même `id_livraison`

Un même programme (`id_livraison`) devrait avoir un seul `mode_real` ; sinon, l'agrégation par mode serait faussée.


```python
incoherent = df.groupby('id_livraison')['mode_real'].nunique()
n_incoherent = (incoherent > 1).sum()
print(f"Programmes avec mode_real incohérent entre leurs lignes : {n_incoherent}")

ids_incoherents = incoherent[incoherent > 1].index
if n_incoherent > 0:
    print(df[df['id_livraison'].isin(ids_incoherents)]
          [['id_livraison', 'adresse_programme', 'annee', 'mode_real']])
```

```python
# Test plus robuste : comparer le début de l'adresse (moins sensible aux petites
# variations de texte) plutôt que l'égalité stricte
for id_ in ids_incoherents:
    sub = df[df['id_livraison'] == id_]
    adresses = sub['adresse_programme'].unique()
    prefixes = set(a.strip().upper()[:15] for a in adresses)
    same_prefix = len(prefixes) == 1
    flag = "OK (adresse similaire)" if same_prefix else "SUSPECT (adresses vraiment différentes)"
    print(f"id_livraison {id_} — {flag}")
    if not same_prefix:
        for a in adresses:
            print(f"   -> {a}")
```

## 10. `nature_programme` × `mode_real` — facteur de confusion

Vérifier si `nature_programme` influence la distribution de `mode_real`, ce qui justifierait de restreindre l'analyse principale au logement familial.


```python
print(df['nature_programme'].value_counts())

print("\nnature_programme x mode_real (pourcentage par ligne) :")
ct = pd.crosstab(df['nature_programme'], df['mode_real'], normalize='index')
print((ct * 100).round(1))
```

```python
print(df.groupby('mode_real')[['nb_plai', 'nb_plus', 'nb_pluscd', 'nb_pls']]
        .agg(['sum', 'mean', 'min']))
```

## 11. Vérification — cohérence `nb_logmt_total` vs sous-catégories

`nb_logmt_total` doit correspondre exactement à la somme de `nb_plai` + `nb_plus` + `nb_pluscd` + `nb_pls`.


```python
df['sum_categories'] = df[['nb_plai', 'nb_plus', 'nb_pluscd', 'nb_pls']].sum(axis=1)
mismatch = df[df['sum_categories'] != df['nb_logmt_total']]
print(f"Lignes où le total ne correspond pas à la somme des sous-catégories : {len(mismatch)}")

if len(mismatch) > 0:
    print(mismatch[['id_livraison', 'adresse_programme', 'annee', 'mode_real',
                     'nb_logmt_total', 'sum_categories']])
```

## Conclusions — Data Understanding

- **Négatifs** (`nb_plai`/`nb_plus`/`nb_logmt_total` < 0) : quelques lignes isolées, 
  probablement des corrections de déclarations antérieures. Impact négligeable 
  (2 lignes / 4174). Décision : exclues du dataset nettoyé.

- **id_livraison dupliqués** : NE SONT PAS des doublons. Ils représentent des tranches 
  de financement ou des phases de livraison distinctes d'un même programme. 
  Décision : conservées telles quelles, id_livraison n'est pas une clé unique par ligne.

- **id_livraison avec mode_real incohérent entre ses lignes** : 7 cas détectés sur le 
  dataset brut (10 lignes). Test de similarité d'adresse (préfixe normalisé) : 
  6 cas sont des tranches/phases légitimes du même programme (même adresse, à des 
  variations d'écriture près). 1 cas (id_livraison 2002196) présente deux adresses 
  réellement différentes (5-7 RUE DU SAHEL vs 1 RUE PAUL CRAMPEL, années 2012/2024), 
  suggérant une réutilisation de l'identifiant plutôt qu'un même programme. 
  Comme l'analyse se fait au niveau ligne (chaque ligne porte son propre mode_real 
  et ses comptes de financement), ceci n'affecte pas l'agrégation par mode_real. 
  Aucune action de nettoyage nécessaire ; documenté pour ne pas traiter id_livraison 
  comme identifiant de projet fiable.

- **bs = "(vide)"** : placeholder de valeur manquante (109 lignes), non utilisé dans 
  l'analyse principale. Pas d'action nécessaire.

- **Outlier nb_logmt_total = 1332** : légitime, cohérent avec les gros programmes 
  d'acquisition conventionnement (rachat de parc immobilier existant, ex. Porte de 
  Montreuil). Conservé.

- **nature_programme fortement corrélé à mode_real** (ex : résidences étudiantes 
  81% en acquisition réhabilitation, EHPAD 92% en acquisition conventionnement) : 
  facteur de confusion potentiel pour l'analyse principale. Décision : analyse 
  centrée sur `nature_programme == "logement familial"` (3551 programmes), avec 
  vérification de robustesse sur le dataset complet.

- **Cohérence nb_logmt_total / sous-catégories** : vérifié que `nb_logmt_total` 
  correspond exactement à la somme de `nb_plai` + `nb_plus` + `nb_pluscd` + `nb_pls` 
  sur 100% des lignes (0 écart).
